```python
import os

import dotenv
from lakefs_spec import LakeFSFileSystem

dotenv.load_dotenv()

my_credentials = {
    "host": os.getenv("LAKEFS_ENDPOINT"),
    "username": os.getenv("LAKEFS_ACCESS_KEY_ID"),
    "password": os.getenv("LAKEFS_SECRET_ACCESS_KEY"),
}

LAKEFS_REPO = os.getenv("LAKEFS_REPOSITORY")

fs = LakeFSFileSystem(my_credentials)
print(f"Connected successfully! Your repositories: {LAKEFS_REPO}")

local_data_path = "../data/"
```

```python
LAKEFS_REPO
```

```python
dir_list = os.listdir(local_data_path)
for file in dir_list:
    print(file)

raw_data_path = f"lakefs://{LAKEFS_REPO}/main/raw/"

raw_data_list = os.listdir(local_data_path + "/raw")
```

```python
fs.listdir("lakefs://bike-data/main/")
```

```python
for file in dir_list:
    fs.put(
        os.path.join(local_data_path, file),
        f"lakefs://bike-data/main/{os.path.basename(file)}",
    )
```

```python
for file in raw_data_list:
    fs.put(
        os.path.join(local_data_path, "raw", file),
        f"lakefs://bike-data/main/raw/{file}",
    )

fs.listdir("lakefs://bike-data/main/raw/")
```
